```python
import pandas as pd

# Create dataset directly
data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny', 'Rainy', 'Sunny', 'Overcast', 'Rainy'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm', 'Warm', 'Warm', 'Warm', 'Cold'],
    'Humidity': ['Normal', 'High', 'High', 'High', 'Normal', 'Normal', 'High', 'Normal'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong', 'Weak', 'Weak', 'Strong', 'Weak'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool', 'Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast': ['Same', 'Same', 'Change', 'Change', 'Same', 'Same', 'Same', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes', 'Yes', 'No']
}

df = pd.DataFrame(data)

print(df)

X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

hypothesis = None

for i in range(len(y)):
    if y[i] == "Yes":
        if hypothesis is None:
            hypothesis = X[i].copy()
        else:
            for j in range(len(hypothesis)):
                if hypothesis[j] != X[i][j]:
                    hypothesis[j] = "?"

print("\nMost Specific Hypothesis:")
print(hypothesis)
```
