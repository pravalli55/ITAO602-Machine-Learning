```python
import pandas as pd

# Create dataset
data = pd.DataFrame({
    'Sky': ['Sunny','Sunny','Rainy','Sunny','Sunny','Rainy','Sunny','Overcast'],
    'AirTemp': ['Warm','Warm','Cold','Warm','Warm','Warm','Cold','Warm'],
    'Humidity': ['Normal','High','High','High','Normal','Normal','Normal','High'],
    'Wind': ['Strong','Strong','Strong','Strong','Weak','Weak','Strong','Strong'],
    'Water': ['Warm','Warm','Warm','Cool','Warm','Cool','Warm','Warm'],
    'Forecast': ['Same','Same','Change','Change','Same','Change','Same','Same'],
    'EnjoySport': ['Yes','Yes','No','Yes','Yes','No','No','Yes']
})

print(data)

concepts = data.iloc[:, :-1].values
target = data.iloc[:, -1].values

# Candidate Elimination
num_attr = len(concepts[0])

S = ['0'] * num_attr
G = [['?'] * num_attr]

print("\nInitial Specific Hypothesis:")
print(S)

print("\nInitial General Hypothesis:")
print(G)

for i in range(len(target)):

    if target[i] == 'Yes':

        if S == ['0'] * num_attr:
            S = list(concepts[i])

        else:
            for j in range(num_attr):
                if S[j] != concepts[i][j]:
                    S[j] = '?'

    else:

        for j in range(num_attr):
            if S[j] != '?' and S[j] != concepts[i][j]:
                G.append(['?'] * j + [S[j]] + ['?'] * (num_attr-j-1))

print("\nFinal Specific Hypothesis:")
print(S)

print("\nFinal General Hypothesis:")
for h in G:
    print(h)
```
