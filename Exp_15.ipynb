```python
# Iris Flower Classification using Naive Bayes

import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder

# 1. Create dataset

data = {
    "SepalLength": [5.1, 4.9, 7.0, 6.4, 6.3],
    "SepalWidth": [3.5, 3.0, 3.2, 3.2, 3.3],
    "PetalLength": [1.4, 1.4, 4.7, 4.5, 6.0],
    "PetalWidth": [0.2, 0.2, 1.4, 1.5, 2.5],
    "Species": ["setosa", "setosa", "versicolor",
                "versicolor", "virginica"]
}

df = pd.DataFrame(data)

# 2. Display dataset

print("Iris Dataset:")
print(df)

# 3. Separate input and output

X = df[[
    "SepalLength",
    "SepalWidth",
    "PetalLength",
    "PetalWidth"
]]

y = df["Species"]

# 4. Encode target labels

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

# 5. Create Naive Bayes model

model = GaussianNB()

# 6. Train the model

model.fit(X, y_encoded)

# 7. Test with a new flower

new_flower = [[5.0, 3.4, 1.5, 0.2]]

prediction = model.predict(new_flower)

# 8. Convert prediction back to flower name

flower_name = encoder.inverse_transform(prediction)

print("\nNew Flower:")
print("Sepal Length = 5.0")
print("Sepal Width  = 3.4")
print("Petal Length = 1.5")
print("Petal Width  = 0.2")

print("\nPredicted Flower:", flower_name[0])
```
